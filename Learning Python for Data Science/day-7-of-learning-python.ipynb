1. Tuples in Python

**Definition**:
A tuple is an ordered, immutable collection of elements. Tuples are often used for fixed collections of data.

Syntax:

    my_tuple = (element1, element2, element3)

Features:

1. Immutable

2. Ordered

3. Allows duplicates

Operations on Tuples:

```python
# Creating tuples with different types of elements
tuple1 = (1, 2, 3)
tuple2 = ('apple', 'banana', 'cherry')
tuple3 = (1, 'hello', 3.14)
print(tuple1)
print(tuple2)
print(tuple3)
```

```python
# Indexing and Slicing
print(tuple1[0])   # First element
print(tuple1[-1])  # Last element
print(tuple1[1:3]) # Slicing
```

```python
# Unpacking a tuple
a, b, c = tuple1
print(a, b, c)
```

```python
for item in tuple1:
    print(item)
```

```python
# Activity:

# Write a program to create a dictionary of student names and their marks, then print students who scored above 75.

students = {"name": ["Vasu","Aarav","ankit"], "marks":[87,67,78]}

for i,j in students.items():
  if i == 'marks':
    for k in j:
      if k >= 75:
        print(students['name'][students[i].index(k)], students['marks'][students[i].index(k)])


# print(students['marks'].index(87))
```

4. List, Set, and Dictionary Comprehensions

List Comprehension:

    squares = [x**2 for x in range(10)]

Set Comprehension:

    unique_chars = {char for char in "hello world"}

Dictionary Comprehension:

    squares_dict = {x: x**2 for x in range(5)}

```python
list1 = []
for i in range(10):
  list1.append(i**2)

print(list1)
```

```python
# list comprehension

# [expected output , expressions, condition]
list2 = [i**2 for i in range(10)]
print(list2)
```

```python
# with a condition
list2 = [i**2 for i in range(10) if i%2 == 0 if i > 5]
print(list2)
```

```python
# set comprehension
set1 = {i for i in "hello world"}
print(set1)
```

```python
numbers = {x % 3 for x in range(10)}
print(numbers)
```

```python
# dictionary comprehension
squares_dict = {x:x**2 for x in range(5)}
print(squares_dict)
```

### Tuples

#### Given a tuple of integers, numbers = (10, 15, 20, 25, 30), write a program to create a new tuple containing only the odd numbers from the original tuple.

```python
numbers = (10, 15, 20, 25, 30)

odd_numbers = [i for i in numbers if i % 2 != 0 ]

odd_numbers = tuple(odd_numbers)

print(odd_numbers)
```

### Dictionaries

#### 1. Create a dictionary where the keys are the first 5 positive integers and the values are their squares. Write code to delete the key-value pair where the key is 3.

```python
kp = {i: i**2 for i in range(1,6) }

del kp[3]

print(kp)
```

```python
kp = {i:i**2 for i in range(1,6) if i != 3}

print(kp)
```

#### 2. Given a dictionary students = {"Alice": 85, "Bob": 72, "Charlie": 91}, write a program to display only those students who scored less than 80.

```python
students = {"Alice": 85, "Bob": 72, "Charlie": 91}

for stu, marks in students.items():
    if marks < 80:
        print(stu, marks)
```

```python
num = {
    'one' : 1,
    'two' : 2,
    'three' : 3,
    'four' : 4
}

max_key, max_value = 0, 0

for key, value in num.items():
    if value > max_value:
        max_key, max_value = key, value

print(max_key, num[max_key])
```
