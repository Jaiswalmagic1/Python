# Univariate analysis

- Definition: Univariate analysis involves examining one variable at a time. The primary goal is to describe the data and find patterns within a single variable.
- Purpose: Summarizes the data and identifies key metrics like mean, median, mode, variance, and distribution.
- Example: Analyzing the distribution of passengers' ages in the Titanic dataset.


```python
df.age.value_counts()
```

```python
# age distribution

plt.hist(df.age,bins = 10)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution')
plt.show()
```

### conclusion :

- The hightest frequency were the young adults from 20-30 age group, showing that mejority of passengers were in their 20s.
- There is a long tail on the right, meaning that fewer passengers were over 65+ age, this is a typical real life behaviour.
- There is a significant number of passengers were from 0-17 age.

What to look going forward
- how many young adults were travelling in whihch class
- how many survived from which age group.
- how many survived from which class

```python
# Gender distribution
sns.countplot(data = df, x = 'sex')
plt.title('Gender Distribution')
plt.show()
```

### conclusion

Gender
- there are close to 2/3 of the population was male.
- number of males may be higher because in those time male were working on ships

```python
# Embarked ticket distribution
sns.countplot(data = df, x = 'embarked')
plt.title('Embarked Distribution')
plt.show()
```

### conclusion

- 'S' has the highest count (well above 600 passengers), making it the primary port of departure.
- 'C' comes second with about 160–170 passengers.
- 'Q' has the lowest count — roughly under 100 passengers.

# Bivariate Analysis
- Definition: Bivariate analysis examines the relationship between two variables to understand how one variable changes concerning the other.
- Purpose: Identifies associations, correlations, and patterns between two variables.
- Example: Studying the survival rate (survived) by gender (sex) in the Titanic dataset.

```python
# survival rate by gender
sns.barplot(data=df, x = 'sex', y = 'survived')
```

- despite having higher male passengers, less males survived then females, which may me because male were saving females
- male 580 = 19% survived

```python
# analyze relationship beetween age and fare
sns.scatterplot(data = df, x = 'age', y='fare', hue='survived')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age v/s Fare distribution')
plt.show()
```

- There were only passengers who paid the fare more than $500
  
- There is a group of passengers who paid between `$200` to `$290`.

- There is a significantly bigger group who baught tickets between `$50` to `$180`.

- Maximum people baught tickets for less than `$60`.

- There is least suvival for passengers between age 16 to 60 who baught ticket for less than `$30`

```python
# analyze relationship between survival rate vs class
sns.barplot(data = df, x = 'class', y='survived', errorbar=None)
plt.xlabel('Class')
plt.ylabel('Survived')
plt.title('Class v/s Survived distribution')
plt.show()
```

- Maximum people who survived were from first class.
- Least people form third class.

# Multivariate
- Definition: Multivariate analysis involves exploring the relationships among three or more variables simultaneously. It provides insights into how variables interact with one another.
- Purpose: Helps in understanding complex relationships and identifying trends or interactions that aren't apparent in univariate or bivariate analysis.
- Example: Analyzing survival rates across gender (sex), class (class), and embarkation point (embarked) in the Titanic dataset.

```python
# Analyze survival rate by gender and class.

sns.catplot(data = df, x='sex', col = 'class', hue = 'survived', kind = 'count')
plt.suptitle('Survival rate by gender and class', y = 1.03)
plt.show()
```

```python
sns.catplot(data = df, x='sex', col = 'survived', hue = 'class', kind = 'count')
plt.suptitle('Survival rate by gender and class', y = 1.03)
plt.show()
```

- Maximum casualty happened in Third class. where males did no survive.
