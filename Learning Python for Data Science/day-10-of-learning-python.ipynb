# NumPy: Numerical Python

NumPy (Numerical Python) is a Python library used for numerical and scientific computing. It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on them efficiently.

## Why Learn NumPy?

NumPy is essential for anyone working in data analysis, machine learning, or scientific computing because:

- It allows efficient handling of large datasets.
- It forms the foundation for libraries like pandas, scikit-learn, TensorFlow, and PyTorch.
- It supports vectorized operations, enabling faster computations compared to Python loops.

````python
# Importing NumPy
import numpy as np
# Creating a 1D array
arr = np.array([1,2,3,4,5])

print("NumPy array: ",arr)
````

### Creating Arrays

NumPy provides several methods to create arrays efficiently:

```python
# 2D array using nested lists
arr_2d = np.array([[1, 2], [3, 4]])
print(arr_2d)

# Array of zeros
zeros = np.zeros((2, 3))
print(zeros)

# Array of ones
ones = np.ones((3, 3))
print(ones)

# Array with a range of numbers
range_array = np.arange(0, 10, 2)  # 0 to 10, step 2
print(range_array)
```

````python
# Example: Creating Arrays
# 2D array using nested lists
arr_2d = np.array([[1,2,3],[3,4,5]])
print(arr_2d)

# Array of zeros
zeros = np.zeros((2,3))
print(zeros)

# Array of ones
ones = np.ones((2,3))
print(ones)
ones[1][1]= 5
print(ones)

# Array with a range of numbers
range_array = np.arange(0,10,2)
range_array
````

````python
# we can also define the data type while creating an array, as the default data type is float.
# the first value in shape of array is for rows and second for column.

zeros = np.zeros((2,3), dtype=int)
print(zeros)

ones = np.ones((2,3), dtype=int)
print(ones)
````

````python
range_array = np.arange(0,23,2)
range_array = range_array.reshape((3,4))

range_array
````

````python
a = np.arange()
````

````python
arr_2d = np.array([
    [1,2,3],
    [3,4,5]
])
print(arr_2d)
````

### Array Operations

NumPy allows arithmetic operations on arrays directly:

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Element-wise addition
print(a + b)

# Element-wise multiplication
print(a * b)

# Scalar operations
print(a * 2)
```

````python
# Example: Array Operations
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Element-wise addition
print(a + b)

# Element-wise multiplication
print(a * b)

# Scalar operations
print(a * 2)
````

````python
a = np.array([[1, 2, 3],[1,2,3]])
#b = np.array([[4, 5, 6],[1,2,3]])

print(a)
print(b)
print(a + b)
````

````python
# addition
ad = np.add(a,b)

ad
````

````python
# multiplication
ml = np.multiply(a,b)
ml
````

````python
# multiplication
c = np.array([[ 4, 10, 18],
       [ 4, 10, 18]])

print(c*3)
````

````python
# raise to power
p = np.power(a,2)
p
````

````python
# raise to power
print(p**3)
````
