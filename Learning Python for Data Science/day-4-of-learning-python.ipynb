# Python Loops - Notes & Practice Questions

## 1. Overview of Loops
Loops allow executing a block of code multiple times until a condition is met. Python provides two main types of loops:
- **For Loop** – Used for iterating over sequences like lists, strings, and ranges.
- **While Loop** – Executes as long as a given condition remains `True`.

## 2. For Loop
**Syntax:**
```python
for variable in sequence:
    # Code block
```


**Example 1: Iterating Over a String**


````python
fruits = 'apple'
for letter in fruits:
    print(letter)
````

**Example 2: Using `range()`**


````python
for i in range(5):  # 0,1,2,3,4
    print(i)
````

## 3. While Loop
**Syntax:**
```python
while condition:
    # Code block
```


**Example: Countdown**


````python
i = 5
while i > 0:
    print(i)
    i -= 1
````

## 4. Control Statements
- **break** – Exits the loop immediately.
- **continue** – Skips the current iteration and moves to the next.
- **pass** – Acts as a placeholder and does nothing.


**Example 1: Using `break`**

````python
for i in range(10):
    if i == 6:
        break
    print(i)
````

**Example 2: Using `continue`**


````python
for i in range(10):
    if i == 6:
        continue
    print(i)
````

## 5. Nested Loops
A loop inside another loop is called a **nested loop**.


**Example:**

````python
for i in range(3):
    for j in range(2):
        print(f"i = {i}, j = {j}")
````

### 6. **Use a for loop to check if a number is prime.**

````python
num = int(input('Enter a number: '))

if num <= 1:
    print(f'{num} is not prime.')
else: 
    for i in range(2,num):
        if num % i == 0:
            print(f'{num} is not prime.')
            break
    else:
        print(f'{num} is prime.')
````

This code takes an integer as input and determines whether it is a prime number.

- If the number is ≤ 1, it is directly declared as not prime.
- Otherwise, it checks divisibility by iterating from 2 to num - 1.
    - If the number is divisible by any i, it is not prime, and the loop exits.
    - If no divisors are found, the number is prime.

Prime numbers are those greater than 1 that are only divisible by 1 and themselves.

### 9. **Count the occurrences of each character in a given string using a loop.**

````python
word = 'banana'

char_count = {}

for char in word:
    if char in char_count:
        char_count[char] +=1
    else:
        char_count[char] = 1
        
for char, count in char_count.items():
    print(f'{char}: {count}')
````

To count the occurrences of each character in a given string, I used a dictionary method and initialized an empty dictionary.

- I iterated through each character in the string using a for loop.
- Inside the loop, the conditional statement checks:
    - If the character is already in the dictionary, its count is increased by 1.
    - If the character is not in the dictionary, it is added with an initial count of 1.

### 10. **Find the largest number in a list using a loop.**

````python
num = [21, 52, 16, 30, 91, 1, 6]
largest_num = 0

for i in range(len(num)):
    if num[i] > largest_num:
        largest_num = num[i]
    else:
        continue

print(largest_num)
````

To find the larget number in the list, 
- I iterated through each number in the list.
- Inside the loop, the conditional statement checks:
    - if the number is greater than largest_num, it replaces the num stored in largest_num:
    - If the number is not greater than the largest_num, loop continues.
