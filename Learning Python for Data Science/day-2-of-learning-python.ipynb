### 1. Data Types

**Definition:** Data types define the kind of value a variable holds. Python has several built-in data types for handling different kinds of data.

**Key Data Types to Cover:**

1. Integer (int): Whole numbers without decimals. Example: 5, -3, 42
2. Float (float): Numbers with decimals. Example: 3.14, -1.5
3. String (str): A sequence of characters enclosed in quotes. Example: "Hello", 'Python'
4. Boolean (bool): Represents True or False values. Example: True, False

```python
# Integer
age = 27
print(age)
print(type(age))
```

```python
# float
age = 27.5
print(age)
print(type(age))
```

```python
# string
name = 'hello'
print(name)
print(type(name))
```

```python
# Boolean
sends_email = True
print(sends_email)
print(type(sends_email))

print(sends_email, type(sends_email))
```

### 2. Typecasting

**Definition:**
 Typecasting (or type conversion) is converting one data type to another. This is useful when you need to perform operations between different types, like combining numbers and strings.


**Common Typecasting Functions:**

1. int(): Converts to an integer.
2. float(): Converts to a float.
3. str(): Converts to a string.

```python
# Convert integer to float
age = 27
print(type(age))
print(float(age))
```

```python
# Convert float to integer
age = 27.5
print(type(age),age)
print(type(int(age)), int(age))
```

```python
# Convert number to string
age = 27
print(type(age),age)
print(type(str(age)), str(age))

str(age)
```

```python
age = 27
str(age)
```

### 3. Indexing

**Definition:**
Indexing allows you to access individual characters in a string using their position (index). In Python, indexes start at 0.

```python
# String indexing

name = 'Vishal'

# V I S H A L
# 0 1 2 3 4 5
#-6-5-4-3-2-1


print(name[0])
print(name[-1])
```

### 4. Slicing

**Definition:**
 Slicing allows you to access a range of characters in a string. You specify the start and end indexes, and Python returns the characters in that range.

Syntax: string[start:end] (the end index is not included)

```python
name = "hello world"

name[0:-1]
```

```python
b = "hello world"
print(b[0:5:2])
print(b[0:7:2])
print(b[::-1])
```

**14. Given a string, write a program to count the occurrences of each character without using dictionaries.**

```python
s = input('Enter a string :')

char_count = []
for char in s:
    found = False
    for i in range(len(char_count)):
        if char_count[i][0] == char:
            char_count[i] = (char_count[i][0], char_count[i][1]+1) #value at position 0 remains same and position 1 increases by 1
            found = True
            break
    if not found:
        char_count.append((char, 1))

for char, count in char_count:
    print(f'{char}: {count}')
```

**18. Explain short-circuiting in logical operators with an example.**

Short-circuiting in logical operators means that Python stops evaluating an expression as soon as the result is determined.

```python
x = 5

# 'or' short-circuits if the first condition is True
result = x > 0 or print("This won't be printed")

# 'and' short-circuits if the first condition is False
result = x < 0 and print("This won't be printed")
```

Explanation:
- In `or`, if the first condition is `True`, Python does not evaluate the second condition.
- In `and`, if the first condition is `False`, Python does not evaluate the second condition.
