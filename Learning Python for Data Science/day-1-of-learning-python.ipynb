**Introduction to Python**

Python is a high-level, versatile programming language known for its readability and simplicity. Created in the late 1980s by Guido van Rossum, it has grown into one of the most popular languages for both beginners and experienced programmers due to its vast libraries, dynamic typing, and cross-platform compatibility.   1. **Ease of Use and Readability:** Python’s syntax is clean and easy to understand, making it beginner-friendly yet powerful.

# COMMENTS

In Python, comments are used to add notes or explanations to code without affecting its execution. They help make the code more readable and understandable, especially for complex logic or when working in a team. Comments are ignored by the Python interpreter.

Types of Comments in Python

**1. Single-Line Comments**

These comments begin with the # symbol and continue to the end of the line.
They are typically used to add short explanations or clarify specific lines of code.

```python
# hello this is a comment
```

**2. Multi-Line Comments**

Multi-line comments are often created using triple quotes (''' or """).
While triple quotes are typically used for multi-line strings, they can also serve as multi-line comments in Python, especially when not assigned to a variable.

```python
'''
Hello this is also a comment.
which is a multiline.
'''
```

**Use of Comments**
1. Single-line comments are ideal for brief notes.
2. Multi-line comments are useful for longer descriptions, such as explaining functions, classes, or complex logic.

By using comments, you can make your code easier to understand for yourself and others who may work on it later.

Q14 **Dynamic Type Change:**

- Assign the value 10 to a variable and print its type.
- Change its value to 10.5 and print its type again.
- Finally, assign "Ten" to the same variable and print its type.
- Explain why Python allows this.

```python
variable = 10
print(type(variable))

variable = 10.5
print(type(variable))

variable = 'Ten'
print(type(variable))


print('''Python is dynamically typed, meaning you don’t need to declare a variable’s type explicitly. The type is determined at runtime based on the assigned value. When you assign 10 to a variable, it becomes an integer. If you later assign 10.5, it becomes a float, and if you assign "Ten", it becomes a string. Python allows this because it handles type management internally, unlike statically typed languages (e.g., C, Java) where variables have fixed types.''')
```
