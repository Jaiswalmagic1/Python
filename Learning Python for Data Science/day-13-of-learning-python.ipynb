## 2. Apply Function Example

Apply Function in Pandas

The `apply()` function in Pandas is used to apply a function along an axis of the DataFrame or Series.

It is a highly flexible method that allows you to apply custom or built-in functions to rows or columns.

**Key Points**:

**Element-wise Transformation**:  You can apply a function to modify values in
a column or row.

**Row/Column Operations**:  Use axis=0 to apply the function to columns (default) and axis=1 to apply it to rows.

**Supports Lambda Functions**:  Simplifies applying small, one-off functions.

**Works on Series and DataFrames**:  Can be applied to a single column or the entire DataFrame.

```python
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df
```

```python
df[['A+5', 'B+6']] = df.apply(lambda x : pd.Series([x['A']+5, x['B']+6]),axis = 1)
df
```

```python
df['A=5'] = df['A'].apply(lambda x: x-5)
```

```python
df.rename(columns = {'A=5':'A-5'}, inplace = True)
df
```

## 3. Basics of Regex

Regular expressions (regex) are a sequence of characters that define a search pattern. They are widely used for text matching, extraction, and substitution tasks.

Key Points:

Pattern Matching: Match specific sequences of characters in strings.

Flexible Search: Supports wildcards, ranges, and repetitions.

Functions: Common regex functions include findall(), search(), match(), and sub().

https://www.w3schools.com/python/python_regex.asp

```python
text = "My phone number is a 9876543210 986 047 ."
# Find the phone number

phone_num = re.findall('\d{5}', text)

phone_num
```

```python
text = "My phone number is a 9876543210 986 047 ."
# Find the phone number

phone_num = re.findall('\d+', text)

phone_num
```

```python
# Question 1: Validate Email Addresses
# Identify valid email addresses from a list.
emails = ['alice@example.com', 'bob_at_example.com', 'charlie@example.co.uk']

email  = [i for i in emails if re.match(r'^[\w.-]+@[\w.-]+\.\w+$',i)]
email
```

```python
# Question 2: Extract Hashtags
# Extract all hashtags from a sentence.
text = "I love #Python and #MachineLearning!"
# Extract hashtags

fin = re.findall(r'#\w+',text)
fin
```

```python
# Question 3: Replace Dates with Placeholder
# Replace all dates in the format YYYY-MM-DD with [DATE].
text = "The meeting is on 2024-12-22."
# Replace dates
dat = re.sub(r'\d{4}-\d{2}-\d{2}', '[date]', text)
dat
```

```python
import pandas as pd

# Sample sales data
data = {
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone', 'Tablet'],
    'Units Sold': [5, 10, 3, 2, 7, 4, 8, 6, 3],
    'Unit Price': [1000, 500, 300, 1000, 500, 300, 1000, 500, 300]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate total sales for each row
df['Total Sales'] = df['Units Sold'] * df['Unit Price']

# Pivot table: Total sales by product
pivot_table = df.pivot_table(
    index='Product', 
    values='Total Sales', 
    aggfunc='sum'
)

# Calculate overall total sales
overall_sales = pivot_table['Total Sales'].sum()

# Calculate percentage contribution of each product
pivot_table['Percentage Contribution'] = (pivot_table['Total Sales'] / overall_sales) * 100

# Print the pivot table
print(pivot_table)
```

```python
# Create a pivot table with custom aggregation (e.g., standard deviation of scores by subject).


# Sample student scores data
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie'],
    'Subject': ['Math', 'Math', 'Math', 'English', 'English', 'English', 'Science', 'Science', 'Science'],
    'Score': [88, 92, 85, 76, 81, 79, 91, 89, 94]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create pivot table using standard deviation as aggregation
pivot_std = pd.pivot_table(
    df,
    index='Subject',
    values='Score',
    aggfunc=np.std
)

print(pivot_std)
```

```python
# Use margins in a pivot table to include row and column totals.

# Sample sales data
data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'East', 'South', 'West'],
    'Product': ['Laptop', 'Laptop', 'Phone', 'Tablet', 'Phone', 'Laptop', 'Tablet', 'Phone'],
    'Sales': [1000, 1200, 500, 700, 1500, 1300, 800, 600]
}

# Create DataFrame
df = pd.DataFrame(data)

# Pivot table with margins
pivot_table = pd.pivot_table(
    df,
    index='Region',
    columns='Product',
    values='Sales',
    aggfunc='sum',
    margins=True,
    margins_name='Total'  # Custom name for totals
)

print(pivot_table)
```

```python
# Create a pivot table with a hierarchical index (multi-level rows and columns).

import pandas as pd

# Sample data
data = {
    'Region': ['North', 'North', 'South', 'South', 'East', 'East', 'West', 'West'],
    'Department': ['Sales', 'Tech', 'Sales', 'Tech', 'Sales', 'Tech', 'Sales', 'Tech'],
    'Product': ['Laptop', 'Laptop', 'Phone', 'Phone', 'Tablet', 'Tablet', 'Laptop', 'Phone'],
    'Quarter': ['Q1', 'Q2', 'Q1', 'Q2', 'Q1', 'Q2', 'Q1', 'Q2'],
    'Revenue': [1000, 1500, 900, 1100, 800, 1200, 950, 1050]
}

df = pd.DataFrame(data)

# Create pivot table with hierarchical index
pivot_table = pd.pivot_table(
    df,
    index=['Region', 'Department'],
    columns=['Quarter', 'Product'],
    values='Revenue',
    aggfunc='sum'
)

print(pivot_table)
```

```python
# Filter a pivot table to show only values where the average sales exceed a threshold.

import pandas as pd

# Sample data
data = {
    'Region': ['North', 'North', 'South', 'South', 'East', 'East', 'West', 'West'],
    'Product': ['Laptop', 'Phone', 'Laptop', 'Phone', 'Tablet', 'Tablet', 'Laptop', 'Phone'],
    'Sales': [1200, 800, 1300, 950, 600, 1100, 1250, 900]
}

df = pd.DataFrame(data)

# Create pivot table: average sales by region and product
pivot = pd.pivot_table(
    df,
    index='Region',
    columns='Product',
    values='Sales',
    aggfunc='mean'
)

# Filter: show only values where average sales > 1000
filtered = pivot.where(pivot > 1000)

print(filtered)
```
