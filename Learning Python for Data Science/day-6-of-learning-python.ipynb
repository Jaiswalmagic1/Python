# 2. Lists in Python

**Definition**:

* A list is an ordered collection (data stored in the sequence it is entered) of elements.
Lists can contain elements of different data types.
Key Features:

* Lists are mutable (modifiable).
Elements are indexed (0-based indexing).

**Common Operations**:

* append(): Adds an item to the end of the list.
* insert(index, item): Inserts an item at a specific index.
* pop(): Removes and returns the last element (or an element at a specified index).
* remove(item): Removes the first occurrence of the specified item.
* Slicing: Access subsets using [start:end:step].

**Best Practices**:

1. Use lists when the order of elements is important.
2. Avoid modifying a list while iterating over it; instead, create a copy or use comprehensions.

Syntax:

    my_list = [element1, element2, element3]

# 3. Sets in Python
**Definition**:

A set is an unordered collection* of unique elements.
Sets do not allow duplicate entries.

-* Elements will not remain in the order of insertion.

**Key Features**:

* Sets are mutable.
* Elements must be immutable (e.g., numbers, strings, tuples).

**Common Operations**:

* add(item): Adds an element to the set.
* remove(item): Removes the specified element (raises an error if not found).
* discard(item): Removes the element (does not raise an error if not found).
* union(): Combines two sets.
* intersection(): Finds common elements between two sets.
* difference(): Finds elements in one set but not the other.
* isdisjoint(): Checks if two sets are disjoint (no common elements).

**Best Practices**:

1. Use sets for membership testing and removing duplicates.
2. Avoid relying on the order of elements.


Syntax

    my_set = {element1, element2, element3}

### Rearrange a sorted list so that the first element is the smallest, the second is the largest, the third is the second smallest, and so on. Example: [1, 2, 3, 4, 5, 6, 7] → Output: [1, 7, 2, 6, 3, 5, 4]

```python
lis = [1, 2, 3, 4, 5, 6, 7] 
output = []

while len(lis)>1:
    min_val = min(lis)
    max_val = max(lis)
    
    output.append(min_val)
    output.append(max_val)
    lis.remove(min_val)
    lis.remove(max_val)

if lis:
    output.append(lis[0])
    
output
```

### An element is a majority if it appears more than n/2 times in a list of size n. Find it in O(n) time. Example: [3, 3, 4, 2, 4, 4, 2, 4, 4] → Output: 4

```python
lis =[3, 3, 4, 2, 4, 4, 2, 4, 4]

size = len(lis)
majority = 0
for i in lis:
    seen = []
    if i in seen:
        continue
    else:
        if lis.count(i) > size / 2:
            majority = i
            size = i
majority
```

```python
def majority_element(nums):
    candidate, count = None, 0

    # Step 1: Find candidate
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)

    # Step 2: Verify candidate
    if nums.count(candidate) > len(nums) // 2:
        return candidate
    return None  # No majority element found

# Example usage
lis = [3, 3, 4, 2, 4, 4, 2, 4, 4]
print(majority_element(lis))  # Output: 4
```

### Given two sets of numbers, find all pairs (a, b) where a is from set1, b is from set2, and (a + b) is even. Example: {1, 2, 3} and {4, 5, 6} → Output: {(2, 4), (2, 6), (1, 5), (3, 5)}

```python
set1 = {1, 2, 3}
set2 = {4, 5, 6}

set3 = set()
for a in set1:
    for b in set2:
        if (a + b) % 2 == 0:
            set3.add((a, b))

print(set3)
```
