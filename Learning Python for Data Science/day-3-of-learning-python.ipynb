## **1. String Operations**

## Introduction:


Strings are sequences of characters, and Python provides many ways to work with them. Knowing how to manipulate strings is essential, as strings are frequently used for data handling and formatting.

#### Key String Operations:

    Concatenation (+): Combines two or more strings.

    Repetition (*): Repeats a string a specified number of times.

    String Length (len()): Returns the number of characters in a string.

    Indexing and Slicing: Allows you to access and modify parts of a string.

    String Methods:

     1. upper() and lower(): Converts all characters to uppercase or lowercase.

     2. capitalize(): This makes the first letter capital

     3. replace(old, new): Replaces parts of a string with another string.

     4. find(substring): Finds the first occurrence of a substring.

     5. split(separator): Splits a string into a list based on a separator.

     6. count(substring): counts the occurence of the given substring from the string.

## **2. Conditional Statements**
#### **Introduction**:

Conditional statements are used to make decisions in a program. They evaluate expressions to determine which block of code should execute.

#### **Types of Conditional Statements:**

1. if Statement: Executes a block of code if a condition is true.

2. if-else Statement: Provides an alternative block if the condition is false.

3. if-elif-else Statement: Checks multiple conditions in sequence.

4. Nested if Statements: An if statement within another if to handle multiple layers of conditions.

### 2. Sum of Even Numbers
- Write a program to find the sum of all even numbers from 1 to n using a while loop.

```python
num = int(input('Énter the number'))
sum = 0

while num>0:
    if num % 2 == 0:
        sum += num
    num -= 1

print(sum)
```

### 6. Check Prime Number
- Write a program to check whether a given number is prime or not.

```python
num = int(input('Enter a number : '))

for i in range(2, num):
    if num % i == 0:
        print(num, ' is not prime.')
        break
    else:
        print(num, ' is prime.')
        break
```

### 7. Find the Factorial of a Number
- Write a program to calculate the factorial of a number n.
- Example:
  
 'Input: 5  
 'Output: 120  (since 5! = 5 × 4 × 3 × 2 × 1)

```python
num = int(input('Enter a number : '))
fact = 1
for i in range(1, num+1):
    fact *= i

print(fact)
```

### 8. Fibonacci Series (First N Terms)
- Write a program to print the first N terms of the Fibonacci sequence.
- Example (if n = 6):

    '0, 1, 1, 2, 3, 5 

```python
num = int(input('Enter a number : '))

a, b = 0, 1

f = []

for i in range(num):
    f.append(a)
    a, b = b, a+b

print(f)
```

### 9. Check if a String is a Palindrome
- Write a program to check if a given string is a palindrome.
- Example:

  'Input: madam
  
  'Output: Palindrome  

```python
inp = input('Enter a word')

if inp == inp[::-1]:
    print(inp, ' is a palindrome.')
else:
    print(inp, ' is not a palindrome.')
```

### 10. Find the Greatest Common Divisor (GCD)
- Write a program to find the GCD (Greatest Common Divisor) of two numbers using a loop.
- Example:

  'makefile

  'Input: 12, 18  

  'Output: 6  

```python
num1, num2 = int(input('Enter the first number : ')), int(input('Enter the second number : '))

gcd = []

for i in range(1, min(num1, num2)+1):
    if num1 % i == 0 and num2 % i == 0:
        gcd.append(i)

print(gcd)
```
