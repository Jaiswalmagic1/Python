### **Introduction to Pandas**
**What is Pandas?**
- **Pandas** is a Python library used for data manipulation and analysis.
- It provides two primary data structures:
  1. **Series**: A one-dimensional array, similar to a list or column in Excel.
  2. **DataFrame**: A two-dimensional table, similar to a spreadsheet or SQL table.

**Why use Pandas?**
- Simplifies data cleaning, manipulation, and analysis tasks.
- Provides tools for:
  - Handling missing data.
  - Filtering and slicing datasets.
  - Performing operations like groupby, aggregation, and merging.

A DataFrame is a table of rows and columns. Here's how to create one:

### **Indexing in Pandas**
Indexing allows selecting specific rows and columns.

#### **5. `.loc` (Label-Based Indexing)**
**Definition**: Selects rows/columns using labels (names).
**Syntax**: `df.loc[row_label, column_label]`

```python
# Select row 1 and column 'Name'

df.loc[1,'Name']

# Select rows 0 to 1 and columns 'Name' and 'City'
df.loc[0:1,['Name', 'City']]
```

```python
# How do you filter rows based on a condition using .loc?
df.loc[df['Age']>25,'Name']
```

#### **6. `.iloc` (Position-Based Indexing)**

```python
data = {
    'Student': ['John', 'Alice', 'Bob', 'Diana', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack'],
        'Math': [85, 90, 78, 88, 92, 76, 85, 89, 81, 95],
        'Science': [92, 87, 83, 89, 85, 80, 88, 90, 84, 93],
        'English': [88, 84, 79, 91, 87, 81, 89, 92, 86, 94]
}
```

```python
df = pd.DataFrame(data)
```

```python
# Display the first 5 rows and the last 3 rows of the DataFrame.

df.head()
df.head(5)
df.tail(3)
```

```python
# Find students who scored more than 85 in Math.

df.loc[df['Math']>85]
```

```python
# List all students with an average score (across all subjects) greater than 90.

df[df.iloc[:,1:].mean(axis=1)>90]
```

```python
# Add a Total column that sums up the scores for Math, Science, and English for each student.
df['Total']= df.iloc[:,1:].sum(axis=1)
```

```python
df
```

```python
# Add an Average column that calculates the average score for each student.

df['Average'] = df.iloc[:,1:4].mean(axis=1)
```

```python
df
```

```python
# Sort the DataFrame by Math in descending order:
# Reseting the index to get fresh index

df = df.sort_values(by="Math", ascending=False).reset_index(drop=True)
df
```

```python
# Sort by the Total column in ascending order.
df.sort_values(by='Total')
```

```python
# Group students based on their Math scores:
# Below 80: "Needs Improvement."
# 80-90: "Good."
# Above 90: "Excellent."

# df['Math_feedback'] = [ for ]

def grade(x):
    if x < 80:
        return 'Needs Improvement'
    elif x >90:
        return 'Exellent'
    else:
        return 'Good'
```

```python
df['Math_feedback'] = df['Math'].map(grade)
```

```python
df
```

```python
# Count the number of students in each group.
df.groupby('Math_feedback').count()['Student']
```

# **Pandas Methods Overview**

### **1. DataFrame Creation & Viewing**
- `pd.DataFrame()`: Converts lists, dictionaries, or NumPy arrays into a DataFrame.
- `df.head(n)`: Displays the first `n` rows of the DataFrame (default is 5).
- `df.tail(n)`: Displays the last `n` rows of the DataFrame (default is 5).
- `df.info()`: Provides a summary of the DataFrame, including column names, non-null counts, and data types.
- `df.shape`: Returns the number of rows and columns in the DataFrame.
- `df.columns`: Returns a list of column names.
- `df.index`: Returns the row labels (index).
- `df.describe()`: Returns a statistical summary of numerical columns.

### **2. Indexing & Selection**
- `df['column_name']`: Access a single column.
- `df[['col1', 'col2']]`: Access multiple columns.
- `df.loc[]`: Label-based indexing for rows/columns.
- `df.iloc[]`: Position-based indexing for rows/columns.
- `df.at[]`: Fast access to a single value by label.
- `df.iat[]`: Fast access to a single value by index position.
- `df.xs()`: Access a cross-section of rows/columns.

### **3. Data Manipulation**
- `df.sort_values()`: Sort by one or more columns.
- `df.sort_index()`: Sort by row index.
- `df.drop()`: Drop rows or columns by label.
- `df.rename()`: Rename columns or rows.
- `df.set_index()`: Set a column as the new index.
- `df.reset_index()`: Reset the index and convert it into a column.
- `df.replace()`: Replace specified values with others.
- `df.fillna()`: Fill missing values with a specified value.
- `df.dropna()`: Drop rows or columns with missing values.
- `df.astype()`: Change the data type of a column.
- `df.apply()`: Apply a function along an axis (rows/columns).
- `df.applymap()`: Apply a function to every element in a DataFrame.
- `df.map()`: Map values from a dictionary or function to a Series.

### **4. Aggregation & Grouping**
- `df.groupby()`: Group the DataFrame by one or more columns.
- `df.agg()`: Apply multiple aggregation functions to grouped data.
- `df.transform()`: Apply a function to each group independently.
- `df.pivot_table()`: Create a pivot table for summarization.

### **5. Data Joining & Merging**
- `pd.merge()`: Merge two DataFrames on a common column or index.
- `df.join()`: Join DataFrames by index.
- `df.concat()`: Concatenate DataFrames along rows or columns.

### **6. String Operations**
- `df['column'].str()`: Perform vectorized string operations (e.g., `.lower()`, `.upper()`, `.contains()`, `.replace()`).

### **7. Handling Time Series Data**
- `pd.to_datetime()`: Convert strings or other formats to datetime.
- `df.resample()`: Resample time-series data.
- `df.shift()`: Shift data by a specified period.
- `df.rolling()`: Apply a rolling window operation for statistics like moving averages.

### **8. Missing Data Handling**
- `df.isna()`: Check for missing values.
- `df.notna()`: Check for non-missing values.
- `df.isnull()`: Same as `isna()`.
- `df.notnull()`: Same as `notna()`.

### **9. Numerical Operations**
- `df.sum()`: Sum of values along columns or rows.
- `df.mean()`: Mean of values along columns or rows.
- `df.median()`: Median of values along columns or rows.
- `df.min()`: Minimum value along columns or rows.
- `df.max()`: Maximum value along columns or rows.
- `df.cumsum()`: Cumulative sum.
- `df.cumprod()`: Cumulative product.

### **10. Pivoting**
- `df.pivot()`: Pivot data based on column values.
- `df.melt()`: Unpivot DataFrame from wide to long format.

### **11. Visualization**
- `df.plot()`: Simple plotting (line, bar, etc.) using Matplotlib.
- `df.hist()`: Plot histograms of numeric columns.

### **12. Working with Categories**
- `df['column'].astype('category')`: Convert a column to a categorical data type.
- `df['column'].cat.codes`: Access category codes of a categorical column.

### **13. Performance**
- `df.memory_usage()`: Check memory usage of the DataFrame.
- `df.memory_usage(deep=True)`: Detailed memory usage, especially for object columns.
