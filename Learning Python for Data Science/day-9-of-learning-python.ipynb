```python
# create a dictionary where key are the numbers from 1 to 5 and values are there squares.

diction = {i : i**2 for i in range(1,6)}

diction
```

```python
key = ['a', 'b', 'c']
value = [1,2,3]

diction = { key[i]: value[i] for i in range(len(key))}

diction
```

```python
# filter a dictionary to keep only items where the value is greater than 2.

original_dict = {'a': 1, 'b':2, 'c':3, 'd':4}

f = {i: j for i, j in original_dict.items() if j >2}

f
```
